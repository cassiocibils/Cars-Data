# Análise Exploratória de Dados (EDA) - Anúncios de Vendas de Carros

Este notebook demonstra o uso de **Pandas** e **Plotly Express** para explorar o conjunto de dados `vehicles_us.csv` para o projeto do curso.

```python
import pandas as pd
import plotly.express as px

# Lendo os dados
car_data = pd.read_csv('../vehicles_us.csv')

# Exibir as primeiras linhas
car_data.head()
```

```python
# Informações gerais sobre o conjunto de dados
car_data.info()
```

```python
# Criar um histograma para a coluna 'odometer'
fig_hist = px.histogram(car_data, x="odometer", title="Distribuição de Quilometragem (Odometer)")
fig_hist.show()
```

```python
# Criar um gráfico de dispersão de Preço x Quilometragem
fig_scatter = px.scatter(
    car_data, 
    x="odometer", 
    y="price", 
    color="condition",
    title="Preço x Quilometragem por Condição do Veículo"
)
fig_scatter.show()
```
